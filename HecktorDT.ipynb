```python
# ============================================================
# DECISION TREE — HECKTOR HPV CLASSIFIER
# Shared data preparation | Seed 42 | No SMOTE
# Same validation and evaluation methodology as HANCOCK
# ============================================================

!pip install -q gdown

import os, json, random, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import gdown

from google.colab import drive
from sklearn.model_selection import train_test_split, ParameterGrid
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score, balanced_accuracy_score, precision_score,
    recall_score, f1_score, roc_auc_score, average_precision_score,
    confusion_matrix, classification_report, roc_curve,
    precision_recall_curve
)

warnings.filterwarnings("ignore", category=FutureWarning)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
os.environ["PYTHONHASHSEED"] = str(SEED)

drive.mount("/content/drive")

# ------------------------------------------------------------
# 1. LOAD AND PREPARE DATA
# ------------------------------------------------------------

gdown.download(
    id="1DGDH2e6dEkpdwwtCvH6ds1CwJT2Hdmkv",
    output="HPV2025.xlsx",
    quiet=False
)

BASE = "/content/drive/MyDrive"
OUT_DIR = os.path.join(BASE, "DT_results_HECKTOR")
os.makedirs(OUT_DIR, exist_ok=True)

df = pd.read_excel("HPV2025.xlsx")

df = df.dropna(subset=["HPV Status"]).copy()

# Keep patient ID for saved predictions, but never use it as a predictor.
if "PatientID" in df.columns:
    df = df.rename(columns={"PatientID": "patient_id"})
else:
    df["patient_id"] = np.arange(len(df))

df["HPV_Label"] = pd.to_numeric(
    df["HPV Status"], errors="coerce"
)

df = df.dropna(subset=["HPV_Label"]).copy()
df["HPV_Label"] = df["HPV_Label"].astype(int)

# ------------------------------------------------------------
# SELECT EXACTLY EIGHT FEATURES
# ------------------------------------------------------------

features = [
    "Age",
    "Gender",
    "Tobacco Consumption",
    "Alcohol Consumption",
    "Performance Status",
    "T-stage",
    "N-stage",
    "M-stage"
]

missing_features = [c for c in features if c not in df.columns]
if missing_features:
    raise KeyError(f"Missing predictor columns: {missing_features}")

df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

df = df[["patient_id"] + features + ["HPV_Label"]].copy()

print("Dataset Shape:", df.shape)
print("\nClass Distribution")
print(df["HPV_Label"].value_counts().sort_index())

# ------------------------------------------------------------
# STRATIFIED TRAIN / VALIDATION / TEST SPLIT (65/15/20)
# ------------------------------------------------------------

train_val, test = train_test_split(
    df,
    test_size=0.20,
    random_state=SEED,
    stratify=df["HPV_Label"]
)

train, val = train_test_split(
    train_val,
    test_size=0.1875,
    random_state=SEED,
    stratify=train_val["HPV_Label"]
)

train = train.copy()
val = val.copy()
test = test.copy()

print(
    "\nRows:", len(df),
    "| Train:", len(train),
    "| Validation:", len(val),
    "| Test:", len(test)
)
print(
    "Class counts (train):",
    train["HPV_Label"].value_counts().sort_index().to_dict()
)
print(
    "Class counts (validation):",
    val["HPV_Label"].value_counts().sort_index().to_dict()
)
print(
    "Class counts (test):",
    test["HPV_Label"].value_counts().sort_index().to_dict()
)

# ------------------------------------------------------------
# MISSING VALUES AND ENCODING
# Fit preprocessing on training data only
# ------------------------------------------------------------

numeric_features = ["Age"]
categorical_features = [
    c for c in features if c not in numeric_features
]

try:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    )
except TypeError:
    encoder = OneHotEncoder(
        handle_unknown="ignore",
        sparse=False
    )

preprocessor = ColumnTransformer([
    (
        "num",
        Pipeline([
            ("imputer", SimpleImputer(strategy="median"))
        ]),
        numeric_features
    ),
    (
        "cat",
        Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("onehot", encoder)
        ]),
        categorical_features
    )
])

X_train = preprocessor.fit_transform(train[features])
X_val = preprocessor.transform(val[features])
X_test = preprocessor.transform(test[features])

y_train = train["HPV_Label"].to_numpy()
y_val = val["HPV_Label"].to_numpy()
y_test = test["HPV_Label"].to_numpy()

counts = np.bincount(y_train, minlength=2)
class_weights = {
    i: len(y_train) / (2 * counts[i])
    for i in range(2)
}

print("Number of Features:", len(features))
print("Encoded Features:", X_train.shape[1])
print("Automatic class weights:", class_weights)
print("SMOTE: False")

# ------------------------------------------------------------
# 2. SELECT TREE SETTINGS USING VALIDATION DATA ONLY
# ------------------------------------------------------------

grid = {
    "max_depth": [2, 3, 4, 5, 7, None],
    "min_samples_split": [2, 5, 10, 15],
    "min_samples_leaf": [2, 5, 10],
    "criterion": ["gini", "entropy"]
}

best_model = None
best_threshold = 0.5
best_key = None
tuning_rows = []

for params in ParameterGrid(grid):
    candidate = DecisionTreeClassifier(
        **params,
        class_weight=class_weights,
        random_state=SEED
    )

    candidate.fit(X_train, y_train)

    val_prob = candidate.predict_proba(X_val)[:, 1]
    thresholds = np.linspace(0.10, 0.90, 81)

    scores = [
        balanced_accuracy_score(
            y_val,
            (val_prob >= t).astype(int)
        )
        for t in thresholds
    ]

    max_score = max(scores)

    threshold = min(
        (
            t for t, score in zip(thresholds, scores)
            if score == max_score
        ),
        key=lambda t: abs(t - 0.5)
    )

    val_pred = (val_prob >= threshold).astype(int)

    auc = (
        roc_auc_score(y_val, val_prob)
        if len(np.unique(y_val)) == 2
        else 0.5
    )

    pr_auc = average_precision_score(y_val, val_prob)

    key = (max_score, auc, pr_auc)

    tuning_rows.append({
        **params,
        "validation_balanced_accuracy": max_score,
        "validation_roc_auc": auc,
        "validation_pr_auc": pr_auc,
        "threshold": float(threshold)
    })

    if best_key is None or key > best_key:
        best_key = key
        best_model = candidate
        best_threshold = float(threshold)

pd.DataFrame(tuning_rows).sort_values(
    [
        "validation_balanced_accuracy",
        "validation_roc_auc",
        "validation_pr_auc"
    ],
    ascending=False
).to_csv(
    os.path.join(OUT_DIR, "validation_tuning.csv"),
    index=False
)

# ------------------------------------------------------------
# 3. FINAL TEST EVALUATION AND SAVE OUTPUTS
# ------------------------------------------------------------

test_prob = best_model.predict_proba(X_test)[:, 1]
test_pred = (test_prob >= best_threshold).astype(int)

metrics = {
    "Accuracy": accuracy_score(y_test, test_pred),
    "Balanced_Accuracy": balanced_accuracy_score(y_test, test_pred),
    "Precision": precision_score(y_test, test_pred, zero_division=0),
    "Recall": recall_score(y_test, test_pred, zero_division=0),
    "F1": f1_score(y_test, test_pred, zero_division=0),
    "ROC_AUC": (
        roc_auc_score(y_test, test_prob)
        if len(np.unique(y_test)) == 2
        else np.nan
    ),
    "PR_AUC": average_precision_score(y_test, test_prob),
    "Threshold": best_threshold
}

print("Selected parameters:", best_model.get_params())
print("\nTEST METRICS\n", pd.Series(metrics))
print("\nConfusion matrix:\n", confusion_matrix(y_test, test_pred))
print(
    "\nClassification report:\n",
    classification_report(y_test, test_pred, zero_division=0)
)

pd.DataFrame([metrics]).to_csv(
    os.path.join(OUT_DIR, "metrics.csv"),
    index=False
)

pd.DataFrame({
    "patient_id": test["patient_id"].to_numpy(),
    "y_true": y_test,
    "probability_positive": test_prob,
    "prediction": test_pred
}).to_csv(
    os.path.join(OUT_DIR, "test_predictions.csv"),
    index=False
)

try:
    feature_names = preprocessor.get_feature_names_out()

    importance = pd.DataFrame({
        "feature": feature_names,
        "importance": best_model.feature_importances_
    }).sort_values("importance", ascending=False)

    importance.to_csv(
        os.path.join(OUT_DIR, "feature_importance.csv"),
        index=False
    )

except Exception as e:
    print("Could not export feature names:", e)

with open(os.path.join(OUT_DIR, "settings.json"), "w") as f:
    json.dump({
        "model": "Decision Tree",
        "dataset": "HECKTOR",
        "seed": SEED,
        "features": features,
        "number_of_features": len(features),
        "target_column": "HPV Status",
        "best_parameters": best_model.get_params(),
        "threshold": best_threshold,
        "class_weight": class_weights,
        "smote": False,
        "split": {
            "train": len(train),
            "validation": len(val),
            "test": len(test)
        }
    }, f, indent=2, default=str)

# ------------------------------------------------------------
# ROC / AUC AND PRECISION-RECALL CURVES
# ------------------------------------------------------------

if len(np.unique(y_test)) == 2:
    fpr, tpr, _ = roc_curve(y_test, test_prob)

    plt.figure()
    plt.plot(
        fpr,
        tpr,
        label=f"ROC AUC = {metrics['ROC_AUC']:.3f}"
    )
    plt.plot([0, 1], [0, 1], "--")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title("Decision Tree — Test ROC Curve")
    plt.legend()
    plt.tight_layout()
    plt.savefig(
        os.path.join(OUT_DIR, "roc_curve.png"),
        dpi=150
    )
    plt.show()

    precision, recall, _ = precision_recall_curve(
        y_test, test_prob
    )

    plt.figure()
    plt.plot(
        recall,
        precision,
        label=f"PR AUC = {metrics['PR_AUC']:.3f}"
    )
    plt.xlabel("Recall")
    plt.ylabel("Precision")
    plt.title("Decision Tree — Test Precision-Recall Curve")
    plt.legend()
    plt.tight_layout()
    plt.savefig(
        os.path.join(OUT_DIR, "pr_curve.png"),
        dpi=150
    )
    plt.show()

print("Saved results to:", OUT_DIR)
```

```python

# ============================================================
# DECISION TREE — HECKTOR
# SELECTED GRID-SEARCH MODEL
# ROC + PR CURVES + 95% BOOTSTRAP CONFIDENCE INTERVALS
# ============================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    roc_curve,
    roc_auc_score,
    precision_recall_curve,
    average_precision_score
)

# ------------------------------------------------------------
# 1. VERIFY SELECTED MODEL
# ------------------------------------------------------------

print("SELECTED DECISION TREE MODEL")
print("=" * 60)

print("Hyperparameters:")
print("Criterion:", best_model.get_params()["criterion"])
print("Max depth:", best_model.get_params()["max_depth"])
print("Min samples leaf:", best_model.get_params()["min_samples_leaf"])
print("Min samples split:", best_model.get_params()["min_samples_split"])
print("Threshold:", best_threshold)

# ------------------------------------------------------------
# 2. GENERATE TEST PROBABILITIES
# ------------------------------------------------------------

y_true = np.asarray(y_test).reshape(-1)

# Use the selected grid-search model
test_probabilities = best_model.predict_proba(X_test)[:, 1]

y_prob = np.asarray(test_probabilities).reshape(-1)

assert len(y_true) == len(y_prob), "Prediction length mismatch."

# Calculate test AUC metrics
roc_auc = roc_auc_score(y_true, y_prob)
pr_auc = average_precision_score(y_true, y_prob)

print("\nTEST RESULTS")
print(f"Balanced Accuracy: {balanced_accuracy_score(y_true, (y_prob >= best_threshold).astype(int)):.4f}")
print(f"ROC-AUC: {roc_auc:.4f}")
print(f"PR-AUC: {pr_auc:.4f}")

# Check against results table
assert np.isclose(roc_auc, 0.5688, atol=0.001), \
    "ROC-AUC does not match the selected results table."

assert np.isclose(pr_auc, 0.9085, atol=0.001), \
    "PR-AUC does not match the selected results table."

# ------------------------------------------------------------
# 3. ROC CURVE
# ------------------------------------------------------------

fpr, tpr, roc_thresholds = roc_curve(
    y_true,
    y_prob
)

plt.figure(figsize=(8, 6))

plt.plot(
    fpr,
    tpr,
    linewidth=2.5,
    label=f"Decision Tree (ROC-AUC = {roc_auc:.4f})"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    linewidth=1.5,
    label="Chance"
)

plt.xlabel("False Positive Rate", fontsize=12)
plt.ylabel("True Positive Rate", fontsize=12)

plt.title(
    "Decision Tree — HECKTOR ROC Curve (Test Set)",
    fontsize=14,
    fontweight="bold"
)

plt.legend(loc="lower right")
plt.grid(alpha=0.25)
plt.xlim(0, 1)
plt.ylim(0, 1.02)
plt.tight_layout()

plt.savefig(
    os.path.join(
        OUT_DIR,
        "DT_HECKTOR_ROC_selected_model.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------
# 4. PRECISION-RECALL CURVE
# ------------------------------------------------------------

precision, recall, pr_thresholds = precision_recall_curve(
    y_true,
    y_prob
)

plt.figure(figsize=(8, 6))

plt.plot(
    recall,
    precision,
    linewidth=2.5,
    label=f"Decision Tree (PR-AUC = {pr_auc:.4f})"
)

plt.axhline(
    y=np.mean(y_true),
    linestyle="--",
    linewidth=1.5,
    label="Positive class prevalence"
)

plt.xlabel("Recall", fontsize=12)
plt.ylabel("Precision", fontsize=12)

plt.title(
    "Decision Tree — HECKTOR Precision-Recall Curve (Test Set)",
    fontsize=14,
    fontweight="bold"
)

plt.legend(loc="lower left")
plt.grid(alpha=0.25)
plt.xlim(0, 1)
plt.ylim(0, 1.02)
plt.tight_layout()

plt.savefig(
    os.path.join(
        OUT_DIR,
        "DT_HECKTOR_PR_selected_model.png"
    ),
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ------------------------------------------------------------
# 5. 95% STRATIFIED BOOTSTRAP CONFIDENCE INTERVALS
# ------------------------------------------------------------

N_BOOTSTRAP = 5000
rng = np.random.default_rng(42)

positive_indices = np.where(y_true == 1)[0]
negative_indices = np.where(y_true == 0)[0]

n_positive = len(positive_indices)
n_negative = len(negative_indices)

roc_auc_bootstrap = []
pr_auc_bootstrap = []

for i in range(N_BOOTSTRAP):

    # Resample positive and negative patients separately
    boot_positive = rng.choice(
        positive_indices,
        size=n_positive,
        replace=True
    )

    boot_negative = rng.choice(
        negative_indices,
        size=n_negative,
        replace=True
    )

    boot_indices = np.concatenate([
        boot_positive,
        boot_negative
    ])

    boot_y = y_true[boot_indices]
    boot_prob = y_prob[boot_indices]

    # ROC-AUC
    roc_auc_bootstrap.append(
        roc_auc_score(boot_y, boot_prob)
    )

    # PR-AUC (Average Precision)
    pr_auc_bootstrap.append(
        average_precision_score(boot_y, boot_prob)
    )

# Calculate percentile 95% confidence intervals
roc_ci_lower, roc_ci_upper = np.percentile(
    roc_auc_bootstrap,
    [2.5, 97.5]
)

pr_ci_lower, pr_ci_upper = np.percentile(
    pr_auc_bootstrap,
    [2.5, 97.5]
)

# ------------------------------------------------------------
# 6. DISPLAY FINAL RESULTS
# ------------------------------------------------------------

print("\n" + "=" * 65)
print("DECISION TREE — HECKTOR")
print("95% BOOTSTRAP CONFIDENCE INTERVALS")
print("=" * 65)

print("Model: Decision Tree")
print("Selection: Validation balanced accuracy")
print("Test patients:", len(y_true))
print("HPV positive:", n_positive)
print("HPV negative:", n_negative)
print("Bootstrap iterations:", N_BOOTSTRAP)

print("\nROC-AUC:")
print(f"Point estimate: {roc_auc:.4f}")
print(f"95% CI: [{roc_ci_lower:.4f}, {roc_ci_upper:.4f}]")

print("\nPR-AUC (Average Precision):")
print(f"Point estimate: {pr_auc:.4f}")
print(f"95% CI: [{pr_ci_lower:.4f}, {pr_ci_upper:.4f}]")

print("=" * 65)

# ------------------------------------------------------------
# 7. SAVE CONFIDENCE INTERVAL RESULTS
# ------------------------------------------------------------

ci_results = pd.DataFrame([{
    "Dataset": "HECKTOR",
    "Model": "Decision Tree",
    "Selection_Method": "Validation_Balanced_Accuracy",
    "Criterion": best_model.get_params()["criterion"],
    "Max_Depth": best_model.get_params()["max_depth"],
    "Min_Samples_Leaf": best_model.get_params()["min_samples_leaf"],
    "Min_Samples_Split": best_model.get_params()["min_samples_split"],
    "Threshold": best_threshold,
    "Bootstrap_Iterations": N_BOOTSTRAP,
    "ROC_AUC": roc_auc,
    "ROC_CI_Lower": roc_ci_lower,
    "ROC_CI_Upper": roc_ci_upper,
    "PR_AUC": pr_auc,
    "PR_CI_Lower": pr_ci_lower,
    "PR_CI_Upper": pr_ci_upper
}])

ci_results.to_csv(
    os.path.join(
        OUT_DIR,
        "DT_HECKTOR_confidence_intervals.csv"
    ),
    index=False
)

print("\nAll results saved successfully to:", OUT_DIR)

print("\nFiles generated:")
print("1. DT_HECKTOR_ROC_selected_model.png")
print("2. DT_HECKTOR_PR_selected_model.png")
print("3. DT_HECKTOR_confidence_intervals.csv")
```
